####**TASK P18.2** <font color= orange>**Complete Rosalind Problem #5  Dictionaries**</font>

````python
# 1. Open the data file
# 2. Read the file (a single line of words)
file_path = "/content/drive/MyDrive/ColabNotebooks/Module3files/rosalind5.txt"
with open(file_path, "r") as file:
    content = file.read()

# 3. Initiate an empty dictionary
word_count = {}

# 4. Loop over the file getting words one at a time
# Split the content into words. .split() by default splits by whitespace.
words = content.split()

for word in words:
    # 5. Check each word to see if it is already in the dictionary
    # a. if it is in the dictionary, add 1 to the value
    # b. if (else) it is not in the dictionary, make it a new key with value = 1
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

# 6. After the loop: Print the key, value pairs for the whole dictionary
for key, value in word_count.items():
    print(key, value)
````

####**TASK BP21.2** <font color=orange> **Complete Rosalind #7 Transcribing DNA into RNA**</font>

````python
#make sure to run the  bio.seq import seq cell beforehand
file_path = "/content/drive/MyDrive/ColabNotebooks/Module3files/rosalindDNA2RNA.txt"

# Open the file and read the DNA sequence
with open(file_path, "r") as file:
    dna = file.read().strip()

# Create a Biopython sequence
my_sequence = Seq(dna)

# Find the complement of the DNA sequence
complement_sequence = my_sequence.complement()

# Transcribe the DNA into RNA
rna = complement_sequence.transcribe()

# Print the RNA sequence
print(rna)
````

####**TASK BP21.2** <font color=orange> **Complete Rosalind #8  Translating RNA into Protein**</font>

````python

file_path = "/content/drive/MyDrive/ColabNotebooks/Module3files/RNA2pro.txt"

# Open the file and read the DNA sequence
with open(file_path, "r") as file:
    dna = file.read().strip()

# Create a Biopython sequence
my_sequence = Seq(dna)

# Transcribe DNA to RNA
rna = my_sequence.transcribe()

# Translate RNA to protein
protein = rna.translate()

# Print the RNA and protein sequences
print("RNA:", rna)
print("Protein:", protein)
````
