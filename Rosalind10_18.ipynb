# Rosalind #10

```python
def pattern_count(text, pattern):
    count = 0
    k = len(pattern)

    # Slide across the text one position at a time
    for i in range(len(text) - k + 1):
        if text[i:i + k] == pattern:
            count += 1

    return count


# Put the path to input file here:
file_path = "/content/drive/MyDrive/ColabNotebooks/Untitled Folder/rosalind10.txt"

# Open the file and read the two lines
with open(file_path, "r") as file:

    #Read the first line aka the DNA string
    text = file.readline().strip()

    #reads the second line, which is the pattern I want
    pattern = file.readline().strip()

#Pattern_count goes through the DNA one position at a time and couns matches
result = pattern_count(text, pattern)

#prints the final number that Rosalind
print(result)
```

# Rosalind #11

```python
# Put the path to input file here
file_path = "/content/drive/MyDrive/ColabNotebooks/R10-18/rosalind11.txt"


# Read the input file
#read first line
with open(file_path, "r") as file:
    text = file.readline().strip()
    #read the second line, convert it to a number (so we know how long of a string to look for)
    k = int(file.readline().strip())


# Dictionary to keep track of how many times
# each k-mer appears
counts = {}

# Look at every k-mer in the DNA string
#For the text we want "#x" (text=, k=)
#len(text) -k + 1 = length of string in text minus k plus 1 (find length then move one over)
for i in range(len(text) - k + 1):
    pattern = text[i:i + k]

    # If we've seen this k-mer before, add 1
    # Otherwise, start its count at 1
    if pattern in counts:
        counts[pattern] += 1
    else:
        counts[pattern] = 1


# Find the highest count
max_count = max(counts.values())


# Find all k-mers that have that highest count
#This creates an empty list where it will store the answers
most_frequent = []

#loop checks highest count 2==2 to another highest count then lists
for pattern in counts:
    if counts[pattern] == max_count:
        most_frequent.append(pattern)


# Print the answer
print(" ".join(most_frequent))
```

# Rosalind #12

```python
# Placeholder: change this to the location of your own input file
file_path = "/content/drive/MyDrive/ColabNotebooks/R10-18/rosalind12.txt"

with open(file_path, "r") as file:

    # First line = the DNA string (Genome)
    Genome = file.readline().strip()

    # Second line contains k, L, and t
    # Example: 5 75 4
    k, L, t = map(int, file.readline().split())


# Set to store the k-mers that form clumps
# We use a set because we only want each k-mer to appear once in our final answer.
clumps = set()

# Look at every possible window of length L
# L = length of the window
for i in range(len(Genome) - L + 1):

    # Take an L-character window from the Genome
    window = Genome[i:i + L]

    # Count how many times each k-mer appears inside this window
    counts = {}

    # A window of length L contains L - k + 1 possible k-mers.
    # Example: L = 75, k = 5
    # 75 - 5 + 1 = 71 possible 5-mers
    for j in range(L - k + 1):

        # Extract a k-mer
        # k = length of the pattern
        pattern = window[j:j + k]


        # If we have seen this k-mer before, increase its count by 1.
        if pattern in counts:
            counts[pattern] += 1

        # If this is the first time we have seen it, start its count at 1.
        else:
            counts[pattern] = 1


    # Check which k-mers occur at least t times
    for pattern in counts:

        # t = minimum number of times the pattern must appear in the L-length window.
        # Example: t = 4
        # If a pattern appears 4, 5, 6, etc. times, it forms a clump.
        if counts[pattern] >= t:

            # Add the k-mer to our set of answers.
            # A set automatically prevents duplicates.
            clumps.add(pattern)

print(" ".join(clumps))
# order not guaranteed
```

# Rosalind #13

```python
#skew (in this case "G" - "C")
file_path = "/content/drive/MyDrive/ColabNotebooks/R10-18/rosalind13.txt"

with open(file_path, "r") as file:

    # Read the DNA string
    Genome = file.readline().strip()

# Keep track of the current skew
# G = +1
# C = -1
skew = 0

# Keep track of the minimum skew we've seen
min_skew = 0

# Keep track of the positions where the minimum skew occurs
positions = []

# Look at every nucleotide in the Genome
for i in range(len(Genome)):

    # Get the current nucleotide
    nucleotide = Genome[i]

    # Update the skew
    if nucleotide == "G":

        # G increases skew by 1
        skew += 1

   # elif == else if
    elif nucleotide == "C":

        # C decreases skew by 1
        skew -= 1


    # Check whether we have found a new minimum
    if skew < min_skew:

        # We found a smaller skew
        min_skew = skew

        # This position is now the only position with the minimum skew
        positions = [i + 1]

    # Check whether this position ties the minimum
    elif skew == min_skew:

        # Add this position to our answer
        positions.append(i + 1)


# Print all positions with the minimum skew

print(" ".join(map(str, positions)))
```

# Rosalind #14

```python
#Hamming distance is the # of positions where 2 dif strings have dif characters

file_path = "/content/drive/MyDrive/ColabNotebooks/R10-18/rosalind14.txt"

with open(file_path, "r") as file:

    # Read the first DNA string
    string1 = file.readline().strip()

    # Read the second DNA string
    string2 = file.readline().strip()

# Keep track of the number of mismatches
hamming_distance = 0

# Compare the two strings one position at a time
for i in range(len(string1)):

    # If the characters at this position are different,increase the Hamming distance by 1.
    if string1[i] != string2[i]:

        hamming_distance += 1

print(hamming_distance)
```

# Rosalind #15

```python
file_path = "/content/drive/MyDrive/ColabNotebooks/R10-18/rosalind15.txt"

with open(file_path, "r") as file:

    # Read the DNA pattern
    Pattern = file.readline().strip()

    # Read d
    d = int(file.readline().strip())


# DNA letters we can use
nucleotides = "ACGT"


# Store all neighbors
neighbors = set()


# Start with the original pattern
# The original pattern has Hamming distance 0 from itself, so it is included when d >= 0.
neighbors.add(Pattern)

# Change each position one at a time
for i in range(len(Pattern)):

    # Try replacing the current character with each of the four DNA nucleotides.
    for nucleotide in nucleotides:

        # Don't replace a character with itself.
        if nucleotide != Pattern[i]:

            # Create a new string with the character at position i replaced.
            neighbor = Pattern[:i] + nucleotide + Pattern[i + 1:]

            # Add the new neighbor to our set.
            neighbors.add(neighbor)


print("\n".join(neighbors))
```

# Rosalind #16

```python
#Find all approximate occurrences of a pattern in a string.
#Given: Strings Pattern and Text along with an integer d.
#Return: All starting positions where Pattern appears as a substring of Text with at most d mismatches.
file_path = "/content/drive/MyDrive/ColabNotebooks/R10-18/rosalind16.txt"

with open(file_path, "r") as file:

    # First line = Pattern
    Pattern = file.readline().strip()

    # Second line = Text
    Text = file.readline().strip()

    # Third line = d
    d = int(file.readline().strip())


# Store the starting positions where we find an approximate match
positions = []

# Look at every possible substring of Text that is the same length as Pattern
for i in range(len(Text) - len(Pattern) + 1):

    # Take a substring from Text that is the same length as Pattern.
    substring = Text[i:i + len(Pattern)]

    # Calculate the Hamming distance
    mismatches = 0

    # Compare Pattern and substring one character at a time.
    for j in range(len(Pattern)):

        # If the characters don't match, we have a mismatch.
        if Pattern[j] != substring[j]:

            mismatches += 1

    # Check if the number of mismatches is small enough
    if mismatches <= d:

        # If there are d or fewer mismatches, save the starting position.
        positions.append(i)

# Print all starting positions
print(" ".join(map(str, positions)))
```

# Rosalind #17

```python
file_path = "/content/drive/MyDrive/ColabNotebooks/R10-18/rosalind17.txt"

with open(file_path, "r") as file:

    # First line = DNA string
    Text = file.readline().strip()

    # Second line = k and d
    k, d = map(int, file.readline().split())

# DNA letters
nucleotides = "ACGT"

# Function to generate all neighbors within d mismatches
def generate_neighbors(pattern, d):

    # Start with the original pattern. It has 0 mismatches with itself.
    neighbors = {pattern}

    # Keep generating new neighbors
    for _ in range(d):

        # Make a new set for the next round
        new_neighbors = set(neighbors)

        # Look at every pattern we already have
        for current in neighbors:

            # Change each position
            for i in range(len(current)):

                # Try each DNA nucleotide
                for nucleotide in nucleotides:

                    # Don't replace a letter with itself
                    if nucleotide != current[i]:

                        # Create a new string with one character changed
                        neighbor = (
                            current[:i]
                            + nucleotide
                            + current[i + 1:]
                        )

                        # Add it to our set
                        new_neighbors.add(neighbor)


        # Update neighbors
        neighbors = new_neighbors


    return neighbors

# Dictionary to count approximate occurrences
counts = {}

# Look at every k-mer in Text
for i in range(len(Text) - k + 1):

    # Get a k-mer from Text
    pattern = Text[i:i + k]

    # Find every k-mer within d mismatches
    neighbors = generate_neighbors(pattern, d)

    # Count each neighbor
    for neighbor in neighbors:

        if neighbor in counts:
            counts[neighbor] += 1

        else:
            counts[neighbor] = 1

# Find the highest count
max_count = max(counts.values())

# Find all most frequent k-mers
most_frequent = []

for pattern in counts:

    if counts[pattern] == max_count:

        most_frequent.append(pattern)


print(" ".join(most_frequent))
```

```python
#Reverse Complement means
#A>T and C>G but then reverse the complement string
```

```python
file_path = "/content/drive/MyDrive/ColabNotebooks/R10-18/rosalind18.txt"

with open(file_path, "r") as file:
    Text = file.readline().strip()
    k, d = map(int, file.readline().split())

# Find reverse complement
def reverse_complement(pattern):
    complement = {
        "A": "T",
        "T": "A",
        "C": "G",
        "G": "C"
    }

    result = ""

    for nucleotide in pattern[::-1]:
        result += complement[nucleotide]

    return result

# Find all strings within d mismatches
def neighbors(pattern, d):
    result = {pattern}

    for _ in range(d):
        new = set(result)

        for word in result:
            for i in range(len(word)):
                for nucleotide in "ACGT":

                    if nucleotide != word[i]:
                        neighbor = word[:i] + nucleotide + word[i + 1:]
                        new.add(neighbor)

        result = new

    return result

# Count approximate matches
counts = {}

for i in range(len(Text) - k + 1):

    pattern = Text[i:i + k]

    # Count neighbors of the pattern
    for word in neighbors(pattern, d):

        if word in counts:
            counts[word] += 1
        else:
            counts[word] = 1

    # Count neighbors of the reverse complement
    reverse = reverse_complement(pattern)

    for word in neighbors(reverse, d):

        if word in counts:
            counts[word] += 1
        else:
            counts[word] = 1

# Find the highest count
max_count = max(counts.values())

# Find all k-mers with the highest count
answer = []

for word in counts:

    if counts[word] == max_count:
        answer.append(word)

print(" ".join(answer))
```
