# Rosalind #18

```python
file_path = "/content/drive/MyDrive/ColabNotebooks/R10-18/rosalind18.txt"

with open(file_path, "r") as file:
    Text = file.readline().strip()
    k, d = map(int, file.readline().split())

# Find reverse complement
def reverse_complement(pattern):
    complement = {
        "A": "T",
        "T": "A",
        "C": "G",
        "G": "C"
    }

    result = ""

    for nucleotide in pattern[::-1]:
        result += complement[nucleotide]

    return result

# Find all strings within d mismatches
def neighbors(pattern, d):
    result = {pattern}

    for _ in range(d):
        new = set(result)

        for word in result:
            for i in range(len(word)):
                for nucleotide in "ACGT":

                    if nucleotide != word[i]:
                        neighbor = word[:i] + nucleotide + word[i + 1:]
                        new.add(neighbor)

        result = new

    return result

# Count approximate matches
counts = {}

for i in range(len(Text) - k + 1):

    pattern = Text[i:i + k]

    # Count neighbors of the pattern
    for word in neighbors(pattern, d):

        if word in counts:
            counts[word] += 1
        else:
            counts[word] = 1

    # Count neighbors of the reverse complement
    reverse = reverse_complement(pattern)

    for word in neighbors(reverse, d):

        if word in counts:
            counts[word] += 1
        else:
            counts[word] = 1

# Find the highest count
max_count = max(counts.values())

# Find all k-mers with the highest count
answer = []

for word in counts:

    if counts[word] == max_count:
        answer.append(word)

print(" ".join(answer))
```
